```python
import uproot
import numpy as np
import matplotlib.pyplot as plt

# Load both ROOT files
silicon = uproot.open("/home/sally/Desktop/itek/senstech-detector-sim/data/modules.root")
cdte = uproot.open("/home/sally/Desktop/itek/senstech-detector-sim/data/modules.root")
# ↑ update the cdte path once you find it from the find command

# ── Charge spectrum comparison ────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, f, label, color in zip(
    axes,
    [silicon, cdte],
    ["Silicon 300μm", "CdTe 1000μm"],
    ["#0D9488", "#0369A1"]
):
    pixel_charge = f["DetectorHistogrammer/medipix/charge/pixel_charge;1"]
    vals = pixel_charge.to_numpy()
    ax.bar(vals[1][:-1], vals[0],
           width=np.diff(vals[1]),
           color=color, alpha=0.8)
    ax.set_title(f"Charge Spectrum — {label}")
    ax.set_xlabel("Pixel Charge (electrons)")
    ax.set_ylabel("Counts")
    total_hits = vals[0].sum()
    ax.text(0.97, 0.95, f"Total hits: {total_hits:.0f}",
            transform=ax.transAxes,
            ha='right', va='top',
            fontsize=11, color=color,
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.suptitle("Silicon vs CdTe — Baseline Comparison @ 60keV", 
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig("baseline_comparison.png", dpi=150)
plt.show()

# ── Summary stats ─────────────────────────────────────────────────────────
print("── Baseline Comparison Summary ─────────────────────")
for f, label, n_events in zip(
    [silicon, cdte],
    ["Silicon", "CdTe"],
    [1000, 5000]
):
    vals = f["DetectorHistogrammer/medipix/charge/pixel_charge;1"].to_numpy()
    hits = vals[0].sum()
    efficiency = (hits / n_events) * 100
    mean_charge = np.average(vals[1][:-1], weights=vals[0]+1e-9)
    print(f"\n{label}:")
    print(f"  Pixel hits detected:  {hits:.0f}")
    print(f"  Detection efficiency: {efficiency:.1f}%")
    print(f"  Mean pixel charge:    {mean_charge:.0f} electrons")
print("─────────────────────────────────────────────────────")
```
